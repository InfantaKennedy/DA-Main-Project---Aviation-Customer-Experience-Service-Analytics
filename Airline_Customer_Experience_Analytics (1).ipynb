## **Dataset Information**


**Dataset:** Skytrax Airline Review Data

**Original Source:** [Mendeley Data – Airline Travel Reviews Data (Skytrax)](https://data.mendeley.com/datasets/vgjk58vf8h/1)

**Original Dataset:** 64,740 rows × 19 columns

**Original Timeline:** 2015–2025

**Dataset Used for Analysis:** 21,984 rows × 19 columns

**Analysis Period:** 2021–2025

**Format:** CSV

The original Skytrax Airline Review Data contains 64,740 airline reviews covering 2015–2025. For this project, the dataset was filtered to include only reviews from 2021–2025, resulting in a working dataset of 21,984 records and 19 columns. The filtered dataset is stored in CSV format and is used for Python-based data cleaning, EDA, visualization, and subsequent Power BI analysis.

**Dataset Column / Features Description:**
- review_date – Date when the review was posted.
- over_all_rating – Overall customer rating from 1–10.
- review_header – Title of the customer review.
- trip_verified – Indicates whether the review is verified.
- review_text – Detailed customer feedback.
- aircraft – Aircraft information; will be removed due to high - missing values and inconsistent entries.
- type_of_traveller – Category of traveller.
- seat_type – Class/type of seat used.
- route – Flight travel route.
- date_flown – Date or period of the flight.
- seat_comfort – Rating for seat comfort.
- cabin_staff_service – Rating for cabin staff service.
- food__beverages – Rating for food and beverages.
- inflight_entertainment – Rating for in-flight entertainment.
- ground_service – Rating for ground services.
- wifi__connectivity – Rating for Wi-Fi/connectivity.
- value_for_money – Rating for value for money.
- recommended – Whether the customer recommends the airline.
- airline_name – Name of the reviewed airline.

```python
# Pivot Table Analysis

# Seat Type × Traveller Type

pivot_table = pd.pivot_table(
    df,
    values="over_all_rating",
    index="seat_type",
    columns="type_of_traveller",
    aggfunc="mean"
)

pivot_table
```

* This gives you a matrix showing:

```python
# Multivariate Analysis

# Visualization 10 - Average Overall Rating by Seat Type and Traveller Type

plt.figure(figsize=(10, 6))

sns.heatmap(
    pivot_table,
    annot=True,
    fmt=".2f"
)

plt.title("Average Overall Rating by Seat Type and Traveller Type")
plt.xlabel("Type of Traveller")
plt.ylabel("Seat Type")

plt.tight_layout()
plt.show()
```
